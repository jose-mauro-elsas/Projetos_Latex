```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4, 300)       # avanço da evaporação (escala ilustrativa)
agua = 100 * np.exp(-1.1 * x)     # % de água restante

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, agua, color="steelblue", linewidth=2)

minerais = [
    ("CaCO₃", 50),
    ("Gesso", 20),
    ("Halita", 10),
    ("Silvita", 3),
]

for nome, percentual in minerais:
    posicao = -np.log(percentual / 100) / 1.1
    ax.scatter(posicao, percentual, color="darkred", zorder=3)
    ax.annotate(nome, (posicao, percentual),
                xytext=(8, 8), textcoords="offset points")

ax.set_xlabel("Progressão da evaporação →")
ax.set_xticks([])
ax.set_ylabel("Volume de salmoura restante (%)")
ax.set_title("Sequência esquemática de precipitação dos evaporitos")
ax.set_ylim(0, 105)
ax.set_yticks(np.arange(0, 101, 10))
ax.grid(True, alpha=0.3)
plt.show()
```

## Concentração de $\mathrm{Mg^{2+}}$ no sal $\mathrm{KCl\cdot MgCl_2\cdot 6H_2O}$

### Enunciado

Uma solução foi preparada dissolvendo-se $5{,}76\ \mathrm{g}$ de
$\mathrm{KCl\cdot MgCl_2\cdot 6H_2O}$ em água suficiente para perfazer
$2{,}000\ \mathrm{L}$. Calcular:

1. a concentração analítica em quantidade de matéria do sal;
2. a concentração em quantidade de matéria de $\mathrm{Mg^{2+}}$.

### a. Massa molar do sal

Utilizando as massas atômicas fornecidas:

$$
M(\mathrm{KCl\cdot MgCl_2\cdot 6H_2O})
=M(\mathrm K)+3M(\mathrm{Cl})+M(\mathrm{Mg})+6M(\mathrm{H_2O})
$$

$$
M=277{,}8535\ \mathrm{g\,mol^{-1}}
$$

O número de mols do sal é:

$$
n_{\mathrm{sal}}
=\frac{m}{M}
=\frac{5{,}76}{277{,}8535}
=2{,}0730\times10^{-2}\ \mathrm{mol}
$$

A concentração analítica do sal é:

$$
C_{\mathrm{sal}}
=\frac{n_{\mathrm{sal}}}{V}
=\frac{2{,}0730\times10^{-2}}{2{,}000}
=1{,}0365\times10^{-2}\ \mathrm{mol\,L^{-1}}
$$

Considerando os três algarismos significativos da massa utilizada:

$$
\boxed{C_{\mathrm{sal}}=1{,}04\times10^{-2}\ \mathrm{mol\,L^{-1}}}
$$

## b. Concentração de $\mathrm{Mg^{2+}}$

A dissolução do sal pode ser representada por:

$$
\mathrm{KCl\cdot MgCl_2\cdot 6H_2O
\longrightarrow K^+ + Mg^{2+} + 3Cl^- + 6H_2O}
$$

O íon cloreto é $\mathrm{Cl^-}$, e não $\mathrm{Cl^{2-}}$. O número 2 em
$\mathrm{MgCl_2}$ indica a presença de dois íons $\mathrm{Cl^-}$ para cada
íon $\mathrm{Mg^{2+}}$.

A fórmula do sal contém **um átomo de magnésio**. Portanto, a relação
estequiométrica é:

$$
1\ \mathrm{mol\ do\ sal}\longrightarrow1\ \mathrm{mol\ de}\ \mathrm{Mg^{2+}}
$$

Consequentemente:

$$
n_{\mathrm{Mg^{2+}}}=n_{\mathrm{sal}}
=2{,}0730\times10^{-2}\ \mathrm{mol}
$$

Dividindo pelo volume total da solução:

$$
[\mathrm{Mg^{2+}}]
=\frac{n_{\mathrm{Mg^{2+}}}}{V}
=\frac{2{,}0730\times10^{-2}}{2{,}000}
=1{,}0365\times10^{-2}\ \mathrm{mol\,L^{-1}}
$$

Assim:

$$
\boxed{[\mathrm{Mg^{2+}}]=1{,}04\times10^{-2}\ \mathrm{mol\,L^{-1}}}
$$

Neste caso, a concentração de $\mathrm{Mg^{2+}}$ é numericamente igual à
concentração analítica do sal, pois existe exatamente um Mg em cada unidade de
fórmula:

$$
\boxed{[\mathrm{Mg^{2+}}]=C_{\mathrm{sal}}}
$$

Não é necessário dividir novamente pela massa molar do magnésio. A massa molar
de $\mathrm{Mg}$ já foi considerada no cálculo da massa molar do composto.
Ela só seria usada novamente se o problema solicitasse a massa de magnésio.

Saída final, com três algarismos significativos:


Concentração analítica do sal = 1.04e-02 mol/L
²⁺] = 1.04e-02 mol/L


```python
k = 39.0983
cl = 35.453
mg = 24.3050
h = 1.0079
o = 15.9994

kcl_m = k + cl
mgcl2_m = mg + 2 * cl
h2o_6_m = 6 * (2 * h + o)

m_s = kcl_m + mgcl2_m + h2o_6_m
massa = 5.76
volume = 2.000

n_sal = massa / m_s
concentracao_sal = n_sal / volume

# Há 1 mol de Mg²⁺ para cada 1 mol do sal.
n_mg = 1 * n_sal
concentracao_mg = n_mg / volume

print(f'Massa molar do sal = {m_s:.4f} g/mol')
print(f'Quantidade de sal = {n_sal:.4e} mol')
print(f'Concentração analítica do sal = {concentracao_sal:.2e} mol/L')
print(f'[Mg²⁺] = {concentracao_mg:.2e} mol/L')
```

5. Determine os coeficientes de atividade e as atividades para $\mathrm{H^+}$ e $\mathrm{OH^-}$ em uma solução que contém 0,05 mol $\mathrm{L^{-1}}$ HCl e 0,06 mol $\mathrm{L^-1}$ Ca(NO3)2, a 25 °C (αH+ = 9 Å, αOH- = 3 Å).


### Esquema para o cálculo das atividades de $\mathrm{H^+}$ e $\mathrm{OH^-}$

#### 1. Dissociação dos eletrólitos

$$
\mathrm{HCl \rightarrow H^+ + Cl^-}
$$

$$
\mathrm{Ca(NO_3)_2 \rightarrow Ca^{2+} + 2NO_3^-}
$$

#### 2. Determinação das concentrações dos íons

| Origem | Íon | Coeficiente estequiométrico $\nu$ | $C_i$ (mol L$^{-1}$) | $Z_i$ |
|---|---|---:|---:|---:|
| HCl | $\mathrm{H^+}$ | 1 | 0,05 | $+1$ |
| HCl | $\mathrm{Cl^-}$ | 1 | 0,05 | $-1$ |
| $\mathrm{Ca(NO_3)_2}$ | $\mathrm{Ca^{2+}}$ | 1 | 0,06 | $+2$ |
| $\mathrm{Ca(NO_3)_2}$ | $\mathrm{NO_3^-}$ | 2 | 0,12 | $-1$ |

> O coeficiente estequiométrico determina a concentração de cada
> espécie iônica e, consequentemente, influencia a força iônica.

#### 3. Cálculo da força iônica

$$
\mu=\frac{1}{2}\sum_i C_iZ_i^2
$$

O $\mathrm{OH^-}$ não é incluído neste cálculo porque sua concentração
na solução بوت_STA Heights Help Lass is extremelyululo? 
é extremamente pequena em comparação com as concentrações dos demais íons.

$$
\boxed{\mu=0,23\ \mathrm{mol\,L^{-1}}}
$$

#### 4. Cálculo dos coeficientes de atividade

Usar a equação de Debye--Hückel para cada íon:

$$
\log\gamma_i=
-\frac{0,51Z_i^2\sqrt{\mu}}
{1+0,33\alpha_i\sqrt{\mu}}
$$

Para o $\mathrm{H^+}$:

$$
\mu=0,23,
\qquad
Z=+1,
\qquad
\alpha_{\mathrm{H^+}}=9\ \text{\AA}
$$

Para o $\mathrm{OH^-}$:

$$
\mu=0,23,
\qquad
Z=-1,
\qquad
\alpha_{\mathrm{OH^-}}=3\ \text{\AA}
$$

#### 5. Atividade do $\mathrm{H^+}$

$$
a_{\mathrm{H^+}}
=
\gamma_{\mathrm{H^+}}[\mathrm{H^+}]
$$

#### 6. Atividade do $\mathrm{OH^-}$

O $\mathrm{OH^-}$ é proveniente da autoionização da água:

$$
\mathrm{H_2O \rightleftharpoons H^+ + OH^-}
$$

Como o exercício site's fill uses? étoiles
setzung Runaway note activity coefficients are considered, use:

$$
K_w=a_{\mathrm{H^+}}a_{\mathrm{OH^-}}
$$

Portanto:

$$
\boxed{
a_{\mathrm{OH^-}}
=
\frac{K_w}{a_{\mathrm{H^+}}}
}
$$

### Roteiro resumido

$$
\boxed{
\begin{aligned}
&\text{dissociação dos eletrólitos}
\\
&\longrightarrow \text{concentrações e cargas dos íons}
\\
&\longrightarrow \mu
\\
&\longrightarrow \gamma_{\mathrm{H^+}}
\text{ e }
\gamma_{\mathrm{OH^-}}
\\
&\longrightarrow a_{\mathrm{H^+}}
\\
&\longrightarrow a_{\mathrm{OH^-}}
\end{aligned}
}
$$

> O $\mathrm{OH^-}$ não entra no cálculo inicial de $\mu$, mas seu
> coeficiente de atividade é calculado com a força iônica da solução.
> Sua atividade é determinada posteriormente por meio de $K_w$.

### Exemplo 1 — Cálculo da massa molar do carbonato de cálcio

Determine a massa molar do carbonato de cálcio, cuja fórmula química é $\mathrm{CaCO_3}$.

As massas atômicas aproximadas dos elementos são:

* $\mathrm{Ca} = 40{,}078\ \mathrm{u}$;
* $\mathrm{C} = 12{,}0107\ \mathrm{u}$;
* $\mathrm{O} = 15{,}9994\ \mathrm{u}$.

A fórmula $\mathrm{CaCO_3}$ indica que cada unidade do composto contém:

* 1 átomo de cálcio;
* 1 átomo de carbono;
* 3 átomos de oxigênio.

Assim, a massa molar do composto é obtida pela soma das massas molares de todos os átomos presentes em sua fórmula:

$$
\begin{aligned}
M(\mathrm{CaCO_3})
&= 1\cdot M(\mathrm{Ca})
 + 1\cdot M(\mathrm{C})
 + 3\cdot M(\mathrm{O})
\end{aligned}
$$

Substituindo os valores:

$$
\begin{aligned}
M(\mathrm{CaCO_3})
&= 1\cdot 40{,}078
 + 1\cdot 12{,}0107
 + 3\cdot 15{,}9994 \\
&= 40{,}078 + 12{,}0107 + 47{,}9982 \\
&= 100{,}0869\ \mathrm{g\cdot mol^{-1}}
\end{aligned}
$$

Portanto:

$$
\boxed{M(\mathrm{CaCO_3}) = 100{,}0869\ \mathrm{g\cdot mol^{-1}}}
$$

Arredondando para duas casas decimais:

$$
\boxed{M(\mathrm{CaCO_3}) \approx 100{,}09\ \mathrm{g\cdot mol^{-1}}}
$$

Portanto, um mol de carbonato de cálcio possui massa de aproximadamente $100{,}09\ \mathrm{g}$.

```python
c = 40.078
o = 15.9994
ca = 12.0107

massa_molar = ca + c + (3 * o)
print(f"A massa molar do Carbonato de cálcio é: {massa_molar} g/mol")
```

## Exemplo 2 — Cálculo da quantidade de matéria

Quantos mols e milimols de ácido benzoico, $\mathrm{HBz}$, estão contidos em $2{,}00\ \mathrm{g}$ do ácido puro?

Dado:

$$
M(\mathrm{HBz}) = 122{,}1\ \mathrm{g\cdot mol^{-1}}
$$

A quantidade de matéria, representada por $n$, é calculada pela razão entre a massa da substância e sua massa molar:

$$
n = \frac{m}{M}
$$

Substituindo os valores:

$$
n =
\frac{2{,}00\ \mathrm{g}}
{122{,}1\ \mathrm{g\cdot mol^{-1}}}
$$

As unidades de grama se cancelam:

$$
n =
\frac{2{,}00}{122{,}1}\ \mathrm{mol}
$$

$$
n = 0{,}01638\ \mathrm{mol}
$$

Considerando três algarismos significativos:

$$
\boxed{n = 0{,}0164\ \mathrm{mol}}
$$

ou, em notação científica:

$$
\boxed{n = 1{,}64\times 10^{-2}\ \mathrm{mol}}
$$

Para converter mol em milimol, utiliza-se a relação:

$$
1\ \mathrm{mol} = 1000\ \mathrm{mmol}
$$

Assim:

$$
n =
0{,}01638\ \mathrm{mol}
\times
\frac{1000\ \mathrm{mmol}}
{1\ \mathrm{mol}}
$$

$$
\boxed{n = 16{,}4\ \mathrm{mmol}}
$$

Portanto, $2{,}00\ \mathrm{g}$ de ácido benzoico correspondem a aproximadamente:

$$
\boxed{0{,}0164\ \mathrm{mol}
\quad\text{ou}\quad
16{,}4\ \mathrm{mmol}}
$$

```python
x = 2 * 1000 / 122.1
print(f"O número de milimols de ácido benzoico é:{x:.2f}")
print(f"O número de mols de ácido benzoico é:{x/1000:.4f}")
```

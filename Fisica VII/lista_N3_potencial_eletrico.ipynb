$$ 
V_p = k\Bigg(\frac{Q_1}{r_1} + \frac{Q_2}{r_2}\Bigg)

$$k≈8,99×109 Nm2/C^2.$$

**a)** aqui a carga q2 está a 10cm no eixo e o ponto P1 está a 5 cm de cada carga.

## Exercício 3

O gráfico a seguir representa o potencial gerado por uma carga elétrica puntiforme, em função da distância dessa carga aos pontos do campo elétrico. O meio é o vácuo.

A partir do gráfico, determine:

- o valor do potencial elétrico $V_1$;
- a distância $d_2$;
- a carga $Q$ que gera o potencial.




### Desenvolvimento

Neste exercício, o objetivo não é calcular a energia potencial elétrica $U$, mas analisar o **potencial elétrico $V$ produzido por uma carga puntiforme**.

Para uma carga puntiforme no vácuo, o potencial elétrico a uma distância $r$ é dado por:

$$
V = k\frac{Q}{r}
$$

onde:

- $V$ é o potencial elétrico;
- $k$ é a constante eletrostática;
- $Q$ é a carga que produz o potencial;
- $r$ é a distância entre a carga e o ponto considerado.

Como $k$ e $Q$ são constantes para esse problema, podemos reorganizar a expressão:

$$
Vr = kQ
$$

Portanto:

$$
Vr = \text{constante}
$$

Isso significa que o potencial elétrico é **inversamente proporcional à distância**:

$$
V \propto \frac{1}{r}
$$

Por esse motivo, o gráfico não é uma reta. À medida que a distância $r$ aumenta, o potencial $V$ diminui seguindo uma curva.

---

#### 1. Utilizando um ponto conhecido do gráfico

O gráfico fornece um ponto em que:

$$
r = 2{,}0\ \mathrm{m}
$$

e

$$
V = 45\ \mathrm{V}
$$

Como:

$$
Vr = kQ
$$

podemos determinar o valor constante de $kQ$ utilizando esse ponto:

$$
kQ = 45\cdot2
$$

Assim, conhecendo $kQ$, podemos utilizar a mesma relação para os demais pontos da curva.

---

#### 2. Determinação de $V_1$

No gráfico, $V_1$ corresponde à distância:

$$
r = 1{,}0\ \mathrm{m}
$$

Como o produto $Vr$ é constante:

$$
V_1\cdot1{,}0 = kQ
$$

Portanto, o valor de $V_1$ pode ser obtido utilizando o valor de $kQ$ calculado anteriormente.

---

#### 3. Determinação de $d_2$

O gráfico mostra que, na distância $d_2$, o potencial elétrico é:

$$
V = 15\ \mathrm{V}
$$

Novamente:

$$
Vr = kQ
$$

Logo:

$$
15d_2 = kQ
$$

A partir dessa expressão é possível determinar $d_2$.

---

#### 4. Determinação da carga $Q$

Depois de determinar o valor de $kQ$, podemos utilizar:

$$
kQ = Vr
$$

e isolar $Q$:

$$
Q = \frac{Vr}{k}
$$

A constante eletrostática no vácuo é:

$$
k \approx 8{,}99\times10^9\ \mathrm{N\,m^2/C^2}
$$

Assim, utilizando o ponto conhecido do gráfico, pode-se determinar a carga $Q$ responsável pelo potencial elétrico apresentado.

---

### Resumo do raciocínio

O ponto principal do exercício é reconhecer que o gráfico representa a função:

$$
V(r)=k\frac{Q}{r}
$$

Como $kQ$ é constante:

$$
Vr=\text{constante}
$$

Então um único ponto conhecido da curva permite determinar os demais valores pedidos no exercício.

```python
k = 8.99e9

# ponto conhecido do gráfico: V = 45 V e r = 2 m
kq = 45 * 2

# V1 corresponde a r = 1 m
v1 = kq / 1

# d2 corresponde a V = 15 V
d2 = kq / 15

# carga Q
q = kq / k

print(f'O valor de V1 é: {v1:.2f} V')
print(f'O valor de d2 é: {d2:.2f} m')
print(f'O valor de Q é: {q:.3e} C')
```

## Exercício 4

A figura mostra um conjunto de três partículas carregadas. Se a partícula de carga $+q$ é deslocada por uma força externa do ponto A para o ponto D, determine se as grandezas a seguir são **positivas, negativas ou nulas**. Justifique as respostas.

**a)** A variação da energia potencial elétrica.

**b)** O trabalho realizado pela força eletrostática sobre a partícula que foi deslocada.

**c)** O trabalho realizado pela força externa.

**d)** Quais seriam as respostas dos itens (a), (b) e (c) se a partícula fosse deslocada do ponto B para o ponto C?



### Desenvolvimento

A energia potencial elétrica da carga $+q$ depende de sua posição em relação às duas cargas fixas $+Q$.

No ponto $A$, as distâncias da carga $+q$ às duas cargas $+Q$ são $d$ e $4d$. Assim:

$$
U_A =
kqQ\left(\frac{1}{d}+\frac{1}{4d}\right)
$$

No ponto $D$, essas distâncias são invertidas, isto é, $4d$ e $d$:

$$
U_D =
kqQ\left(\frac{1}{4d}+\frac{1}{d}\right)
$$

Portanto:

$$
U_A = U_D
$$

Logo, a variação da energia potencial elétrica no deslocamento de $A$ para $D$ é nula:

$$
\Delta U = U_D-U_A = 0
$$

Em d) $U_B = U_C$ então $\Delta U = 0$, portanto o trabalho da força eletrostática é nulo.

Como $\Delta U=0$, o trabalho realizado pela força também é nulo.

## Exercício 5

São dadas as linhas de campo elétrico e as superfícies equipotenciais de um campo elétrico uniforme. Ver figura.

Determine:

**a)** A intensidade do campo elétrico $E$;

**b)** O potencial elétrico no ponto C;

**c)** O trabalho realizado pela força elétrica sobre uma carga

$$
q=1\ \mu\mathrm{C},
$$

quando ela é deslocada do ponto A até o ponto C.



### Desenvolvimento

### Desenvolvimento

Neste exercício, o campo elétrico é uniforme e a figura fornece superfícies equipotenciais.

Por isso, não devemos usar a expressão:

$$
E = k\frac{q}{r^2}
$$

pois essa fórmula é usada para o campo elétrico produzido por uma carga puntiforme.

Neste caso, devemos usar a relação entre campo elétrico uniforme e diferença de potencial:

$$
E = \frac{|\Delta V|}{d}
$$

---

#### a) Intensidade do campo elétrico

Entre os pontos A e B, a figura informa:

$$
V_A = 100\ \mathrm{V}
$$

$$
V_B = 80\ \mathrm{V}
$$

e a distância entre as superfícies equipotenciais é:

$$
d = 0{,}1\ \mathrm{m}
$$

A diferença de potencial é:

$$
|\Delta V| = |V_B - V_A|
$$

$$
|\Delta V| = |80 - 100|
$$

$$
|\Delta V| = 20\ \mathrm{V}
$$

Logo:

$$
E = \frac{20}{0{,}1}
$$

$$
E = 200\ \mathrm{V/m}
$$

Como:

$$
1\ \mathrm{V/m} = 1\ \mathrm{N/C}
$$

temos:

$$
E = 200\ \mathrm{N/C}
$$

---

#### b) Potencial elétrico no ponto C

O campo elétrico aponta para a direita.

No sentido do campo elétrico, o potencial elétrico diminui.

Do ponto B até o ponto C, a distância é:

$$
d_{BC} = 0{,}2\ \mathrm{m}
$$

Para um campo elétrico uniforme:

$$
\Delta V = -Ed
$$

Portanto:

$$
\Delta V_{BC} = -(200)(0{,}2)
$$

$$
\Delta V_{BC} = -40\ \mathrm{V}
$$

Como:

$$
V_B = 80\ \mathrm{V}
$$

temos:

$$
V_C = V_B + \Delta V_{BC}
$$

$$
V_C = 80 - 40
$$

$$
V_C = 40\ \mathrm{V}
$$

---

#### c) Trabalho realizado pela força elétrica

A carga deslocada é:

$$
q = 1\ \mu\mathrm{C}
$$

Convertendo para coulombs:

$$
q = 1\times10^{-6}\ \mathrm{C}
$$

O trabalho realizado pela força elétrica pode ser calculado por:

$$
W_{\mathrm{el}} = q(V_A - V_C)
$$

Como:

$$
V_A = 100\ \mathrm{V}
$$

e:

$$
V_C = 40\ \mathrm{V}
$$

temos:

$$
W_{\mathrm{el}} =
(1\times10^{-6})(100-40)
$$

$$
W_{\mathrm{el}} =
(1\times10^{-6})(60)
$$

$$
W_{\mathrm{el}} =
6\times10^{-5}\ \mathrm{J}
$$

O trabalho é positivo porque a carga é positiva e seu deslocamento ocorre no sentido do campo elétrico.

---

### Resumo

Para este exercício:

$$
E = \frac{|\Delta V|}{d}
$$

$$
V_C = V_B - Ed_{BC}
$$

$$
W_{\mathrm{el}} = q(V_A - V_C)
$$

O ponto principal é perceber que o campo elétrico já existe e é uniforme. A carga de $1\ \mu\mathrm{C}$ não produz o campo; ela é colocada nesse campo e aparece apenas no cálculo do trabalho.

```python
# item a - intensidade do campo
v_a = 100
v_b = 80
d_ab = 0.1

e = abs(v_b - v_a) / d_ab

# item b - potencial em C
d_bc = 0.2
v_c = v_b - e * d_bc

# item c - trabalho da força elétrica
q = 1e-6
w = q * (v_a - v_c)

print(f'A intensidade do campo é: {e:.2f} N/C')
print(f'O potencial no ponto C é: {v_c:.2f} V')
print(f'O trabalho realizado é: {w:.2e} J')
```

# Formulário — Potencial Elétrico

## 1. Constante eletrostática

No vácuo:

$$
k \approx 8{,}99\times10^9\ \mathrm{N,m^2/C^2}
$$

Também pode ser escrita como:

$$
k=\frac{1}{4\pi\varepsilon_0}
$$

onde $\varepsilon_0$ é a permissividade elétrica do vácuo.

### Aplicabilidade

A constante $k$ aparece nas expressões associadas a **cargas puntiformes**, como força elétrica, campo elétrico e potencial elétrico produzido por uma carga.

---

## 2. Força elétrica — Lei de Coulomb

$$
F=k\frac{|q_1q_2|}{r^2}
$$

### Aplicabilidade

Usada quando queremos determinar a **força de interação entre duas cargas puntiformes**.

* cargas de mesmo sinal $\rightarrow$ repulsão;
* cargas de sinais opostos $\rightarrow$ atração.

### Atenção

Não deve ser usada diretamente quando o exercício pede apenas o potencial elétrico de uma carga.

---

## 3. Campo elétrico produzido por uma carga puntiforme

$$
E=k\frac{|Q|}{r^2}
$$

Na forma vetorial:

$$
\vec{E}=k\frac{Q}{r^2}\hat{r}
$$

### Aplicabilidade

Usada quando uma **carga-fonte $Q$** produz um campo elétrico e queremos calcular a intensidade desse campo a uma distância $r$.

A carga de prova não aparece nessa expressão.

---

## 4. Potencial elétrico produzido por uma carga puntiforme

$$
V=k\frac{Q}{r}
$$

### Aplicabilidade

Usada quando queremos saber o **potencial elétrico em um ponto situado a uma distância $r$ de uma carga puntiforme $Q$**.

O potencial é uma grandeza **escalar**.

Por isso, não precisamos considerar direção ou sentido.

Para uma carga positiva:

$$
Q>0 \Rightarrow V>0
$$

Para uma carga negativa:

$$
Q<0 \Rightarrow V<0
$$

Normalmente, para uma carga puntiforme, considera-se:

$$
V=0
$$

no infinito.

---

## 5. Potencial produzido por várias cargas

Para duas ou mais cargas puntiformes, os potenciais são somados algebricamente:

$$
V_P=
k\frac{Q_1}{r_1}
+
k\frac{Q_2}{r_2}
+\cdots
$$

ou:

$$
V_P=
k\left(
\frac{Q_1}{r_1}
+
\frac{Q_2}{r_2}
+\cdots
\right)
$$

### Aplicabilidade

Usada quando existem **várias cargas-fonte** e queremos determinar o potencial resultante em um ponto $P$.

Como o potencial é escalar, fazemos uma soma comum. Não é necessário decompor vetores.

Foi a relação utilizada no **Exercício 1**.

---

## 6. Relação entre potencial e distância para uma carga puntiforme

Partindo de:

$$
V=k\frac{Q}{r}
$$

podemos escrever:

$$
Vr=kQ
$$

Para uma mesma carga $Q$:

$$
Vr=\text{constante}
$$

Portanto:

$$
V\propto\frac{1}{r}
$$

### Aplicabilidade

Útil quando temos um gráfico de potencial em função da distância.

Se conhecemos um ponto do gráfico, podemos determinar outros pontos usando:

$$
V_1r_1=V_2r_2
$$

Foi a ideia utilizada no **Exercício 3**.

### Atenção

O gráfico de $V$ em função de $r$ não é uma reta, pois:

$$
V\propto\frac{1}{r}
$$

---

## 7. Energia potencial elétrica

$$
U=qV
$$

onde:

* $U$ = energia potencial elétrica;
* $q$ = carga colocada no ponto;
* $V$ = potencial elétrico existente naquele ponto.

### Aplicabilidade

Usada quando conhecemos uma carga $q$ e o potencial elétrico $V$ do ponto onde ela é colocada.

Foi a relação utilizada no **Exercício 2**.

### Atenção

Não confundir:

$$
V=k\frac{Q}{r}
$$

com:

$$
U=qV
$$

A primeira calcula **potencial elétrico**.

A segunda calcula **energia potencial elétrica**.

---

## 8. Variação da energia potencial elétrica

$$
\Delta U=U_f-U_i
$$

Como:

$$
U=qV
$$

também podemos escrever:

$$
\Delta U=q(V_f-V_i)
$$

### Aplicabilidade

Usada quando uma carga se desloca de um ponto inicial para outro ponto final.

Se:

$$
U_f>U_i
$$

então:

$$
\Delta U>0
$$

Se:

$$
U_f<U_i
$$

então:

$$
\Delta U<0
$$

Se:

$$
U_f=U_i
$$

então:

$$
\Delta U=0
$$

Foi utilizada no **Exercício 4**.

---

## 9. Trabalho realizado pela força elétrica

Como a força eletrostática é conservativa:

$$
W_{\mathrm{el}}=-\Delta U
$$

Também podemos escrever:

$$
W_{\mathrm{el}}=q(V_i-V_f)
$$

### Aplicabilidade

Usada para determinar o trabalho realizado **pela própria força elétrica** quando uma carga se desloca entre dois pontos.

Se:

$$
\Delta U<0
$$

então:

$$
W_{\mathrm{el}}>0
$$

Se:

$$
\Delta U>0
$$

então:

$$
W_{\mathrm{el}}<0
$$

Se:

$$
\Delta U=0
$$

então:

$$
W_{\mathrm{el}}=0
$$

---

## 10. Trabalho de uma força externa

Quando o deslocamento é realizado lentamente, sem alteração da energia cinética:

$$
W_{\mathrm{ext}}=\Delta U
$$

Portanto:

$$
W_{\mathrm{ext}}=-W_{\mathrm{el}}
$$

### Aplicabilidade

Usada quando o exercício menciona explicitamente uma **força externa deslocando uma carga** e podemos considerar que não existe variação líquida da energia cinética.

No Exercício 4, como os pontos inicial e final considerados possuíam a mesma energia potencial:

$$
\Delta U=0
$$

logo:

$$
W_{\mathrm{el}}=0
$$

e:

$$
W_{\mathrm{ext}}=0
$$

---

## 11. Campo elétrico uniforme e diferença de potencial

Para um campo elétrico uniforme:

$$
E=\frac{|\Delta V|}{d}
$$

onde $d$ é a distância medida **na direção do campo elétrico**.

Também:

$$
\Delta V=-Ed
$$

quando o deslocamento ocorre na direção do campo.

### Aplicabilidade

Usada quando o exercício fornece:

* um campo elétrico uniforme;
* duas superfícies equipotenciais;
* a diferença de potencial entre elas;
* a distância entre elas.

Foi utilizada no **Exercício 5**.

### Interpretação física

O campo elétrico aponta no sentido de **diminuição do potencial**:

$$
\text{maior }V
\longrightarrow
\text{menor }V
$$

Portanto, no sentido de $\vec E$:

$$
\Delta V<0
$$

---

## 12. Relação entre campo e potencial em campo uniforme

Se conhecemos $E$ e queremos determinar a diferença de potencial entre dois pontos:

$$
\Delta V=-Ed
$$

ou:

$$
V_f=V_i-Ed
$$

### Aplicabilidade

Útil quando conhecemos o potencial em uma superfície equipotencial e queremos determinar o potencial de outra superfície situada a uma determinada distância na direção do campo.

No Exercício 5:

$$
V_C=V_B-Ed_{BC}
$$

---

## 13. Superfícies equipotenciais

Em uma superfície equipotencial:

$$
V=\text{constante}
$$

Portanto, entre dois pontos da mesma superfície:

$$
\Delta V=0
$$

Para uma mesma carga:

$$
\Delta U=q\Delta V=0
$$

e:

$$
W_{\mathrm{el}}=0
$$

### Aplicabilidade

Útil para reconhecer situações em que dois pontos diferentes possuem o **mesmo potencial elétrico**.

As linhas de campo elétrico são perpendiculares às superfícies equipotenciais.

# Como escolher a fórmula

Antes de calcular, identificar **o que o exercício está pedindo**.

| O exercício pede                                         | Relação principal            |          |         |

# Esquema para não confundir as grandezas

Uma **carga-fonte $Q$** produz:

$$
Q
\longrightarrow
\vec E
$$

e também produz:

$$
Q
\longrightarrow
V
$$

Se colocarmos uma carga $q$ nesse ponto:

$$
V+q
\longrightarrow
U
$$

Se essa carga se deslocar entre dois pontos:

$$
\Delta U
\longrightarrow
W_{\mathrm{el}}
$$

Assim:

$$
\boxed{
Q+r \longrightarrow V
}
$$

$$
\boxed{
q+V \longrightarrow U
}
$$

$$
\boxed{
\Delta U \longrightarrow W_{\mathrm{el}}
}
$$

Esse é o encadeamento principal usado nos exercícios da Lista N-3 sobre potencial elétrico.

## 1. Cargas elétricas e Lei de Coulomb

### 1.1 Ideias fundamentais

- A carga elétrica pode ser positiva ou negativa. Cargas de mesmo sinal se repelem; de sinais opostos se atraem.
- Em um sistema isolado, a carga total se conserva: $\sum q_\text{antes}=\sum q_\text{depois}$.
- A carga é quantizada: $q=ne$, com $e=1{,}602\times10^{-19}\ \mathrm C$ e $n$ inteiro.
- Atrito transfere elétrons entre materiais; não cria carga total.
- Em condutores idênticos colocados em contato, a carga total se redistribui igualmente.

### 1.2 Lei de Coulomb

Para duas cargas puntiformes separadas por $r$:

$$F=k\frac{|q_1q_2|}{r^2}.$$

A equação fornece o **módulo**. A direção é a reta que une as cargas; o sentido vem dos sinais. Para várias cargas, aplica-se superposição vetorial.

### Estratégia

1. Converter $\mu\mathrm C$, $\mathrm{nC}$ e $\mathrm{pC}$ para coulombs.
2. Calcular separadamente cada força.
3. Definir um eixo e atribuir sinais ou decompor em componentes.
4. Somar vetorialmente.

## 2. Campo elétrico

### 2.1 Definição e sentido

$$\vec E=\frac{\vec F}{q_0},\qquad \vec F=q_0\vec E.$$

O campo existe independentemente da carga de prova. Para $q_0>0$, força e campo têm o mesmo sentido; para $q_0<0$, sentidos opostos.

### 2.2 Carga pontual e superposição

$$\vec E=k\frac{q}{r^2}\hat r,\qquad
\vec E_\text{resultante}=\sum_i\vec E_i.$$

O campo sai de cargas positivas e entra em cargas negativas. A densidade de linhas representa o módulo. Linhas não se cruzam.

Em coordenadas, para fonte em $\vec r_i$ e ponto $\vec r$:

$$\vec E_i=kq_i\frac{\vec r-\vec r_i}{|\vec r-\vec r_i|^3}.$$

### 2.3 Dipolo

$$\vec p=q\vec d,$$

com $\vec d$ apontando da carga negativa para a positiva.

### 2.4 Distribuições contínuas vistas em aula

- Plano infinito: $E=|\sigma|/(2\varepsilon_0)$ de cada lado.
- Dois planos $+\sigma$ e $-\sigma$: entre eles $E=|\sigma|/\varepsilon_0$; fora, os campos se cancelam.
- Anel de raio $R$, no eixo a uma distância $z$: $E_z=kQz/(z^2+R^2)^{3/2}$.
- Disco de raio $R$: $E_z=(\sigma/2\varepsilon_0)\left(1-z/\sqrt{z^2+R^2}\right)$.

## 3. Potencial elétrico, trabalho e energia

### 3.1 Força conservativa

O trabalho eletrostático depende apenas dos pontos inicial e final:

$$\Delta U=U_f-U_i=-W_\mathrm{el}.$$

### 3.2 Potencial e energia potencial

$$V=\frac Uq,\qquad U=qV,\qquad \Delta U=q\Delta V.$$

O potencial é **escalar**. Para cargas puntiformes:

$$V=k\frac qr,\qquad V_\text{total}=k\sum_i\frac{q_i}{r_i}.$$

Não use módulo em $q_i$: o sinal da carga faz parte do potencial.

### 3.3 Trabalho e campo

$$W_\mathrm{el}=-\Delta U=q(V_i-V_f).$$

Em campo uniforme ao longo de $x$:

$$E=-\frac{\Delta V}{\Delta x}.$$

O campo aponta para potenciais menores. Superfícies equipotenciais são perpendiculares às linhas de campo e o trabalho ao longo delas é nulo.

## 4. Capacitância

### 4.1 Definição

$$C=\frac Q{\Delta V},\qquad Q=C\Delta V.$$

A carga $Q$ é o módulo da carga de uma placa; as placas têm $+Q$ e $-Q$, portanto a carga líquida do capacitor é zero. A unidade é o farad (F).

### 4.2 Placas paralelas

$$C=\varepsilon_0\frac Ad.$$

Com dielétrico preenchendo o espaço:

$$C=\kappa\varepsilon_0\frac Ad=\kappa C_0.$$

A capacitância depende da geometria e do meio, não dos valores momentâneos de $Q$ ou $V$.

### 4.3 Associações

| Associação | Grandeza comum | Equivalente |
|---|---|---|
| Paralelo | mesma $V$ | $C_\mathrm{eq}=\sum C_i$ |
| Série | mesmo $Q$ | $1/C_\mathrm{eq}=\sum 1/C_i$ |

Para dois em série: $C_\mathrm{eq}=C_1C_2/(C_1+C_2)$.

### 4.4 Energia armazenada

$$U=\frac12CV^2=\frac{Q^2}{2C}=\frac12QV.$$

Antes de analisar um dielétrico, pergunte: o capacitor continua ligado à bateria (**V constante**) ou foi desconectado (**Q constante**)?


### Lista 4 — questões V/F

<details><summary><strong>Gabarito comentado</strong></summary>

1. **F** — as placas têm $+q$ e $-q$.  
2. **F** — o dielétrico aumenta $C$ por um fator $\kappa$.  
3. **F** — maior $\kappa$, maior capacitância.  
4. **F** — em série, $Q$ é a mesma; em paralelo, $V$ é a mesma.  
5. **V** — $Q=CV$.  
6. **F** — a unidade é o farad.  
7. **V** — dois capacitores $C$ em série dão $C/2$.  
8. **F** — $C=\varepsilon A/d$ diminui quando $d$ aumenta.  
9. **V** — $Q=(2\,\mu\mathrm F)(10\,\mathrm V)=20\,\mu\mathrm C$.  
10. **F** — na ruptura, o dielétrico passa a conduzir; não se torna isolante.

</details>




- Um capacitor é formado por duas superfícies (placas) separados por um dielétrico. Numa placa fica a carregada positivamente e a outra fica carregada negativamente.  
- A presença de um dielétrico com aumenta a capacitância, uma vez que permite a diminuição da distância.  
- Quanto maior a constante dielétrica, maior a capacitância.  
- A carga em capacitores em série é a soma das cargas, já em paralelo é a média harmônica entre os capacitores.

| `CSER` | `'CE=C1*C2/(C1+C2)'` | Dois capacitores em série |
| `UCAP` | `'U=C*V^2/2'` | Energia do capacitor |

### Observações importantes na HP

- Use `EE` para o módulo do campo, evitando confusão visual com a tecla de expoente `E`.
- Em `POT`, preserve o sinal de `Q`; potencial é escalar e pode ser negativo.
- `ABS` é apropriado em `COUL` e `CAMPO` porque essas equações calculam módulos.
- Para três ou mais capacitores, é mais seguro fazer associações sucessivas de dois em dois do que criar uma fórmula enorme.
- A HP resolve a aritmética, mas não escolhe o sentido do vetor: faça antes um pequeno desenho e defina os eixos.
- Para superposição em duas dimensões, calcule $E_x$ e $E_y$ separadamente e só então use $E=\sqrt{E_x^2+E_y^2}$ e $\theta=\operatorname{atan2}(E_y,E_x)$.

### Duas equações extras úteis para componentes

| Nome | Equação |
|---|---|
| `MOD2` | `'M=SQRT(X^2+Y^2)'` |
| `ANG2` | `'T=ATAN(Y/X)'` |

Para `ANG2`, confira o quadrante no desenho; a forma `ATAN(Y/X)` sozinha não distingue todos os quadrantes.


## 6. Folha de revisão rápida

| Tema | Relação principal | Pergunta de controle |
|---|---|---|
| Coulomb | $F=k|q_1q_2|/r^2$ | atração ou repulsão? |
| Campo | $\vec E=\vec F/q_0$ | qual é o vetor criado pelas fontes? |
| Superposição | $\vec E_R=\sum\vec E_i$ | quais componentes cancelam? |
| Dipolo | $\vec p=q\vec d$ | sentido de $-q$ para $+q$? |
| Potencial | $V=k\sum q_i/r_i$ | mantive os sinais das cargas? |
| Energia | $U=qV$ | o sinal de $q$ foi considerado? |
| Trabalho | $W=q(V_i-V_f)$ | é trabalho do campo ou externo? |
| Capacitor | $Q=CV$ | $Q$ é o módulo em uma placa? |
| Placas | $C=\varepsilon A/d$ | mudou geometria ou dielétrico? |
| Paralelo | mesma $V$ | somar capacitâncias |
| Série | mesmo $Q$ | somar inversos |
| Energia | $U=CV^2/2$ | bateria ligada ($V$) ou desligada ($Q$)? |

### Erros que mais derrubam resultado

- Usar centímetros ou microcoulombs sem converter para SI.
- Somar módulos quando a grandeza é vetorial.
- Colocar módulo nas cargas ao calcular potencial.
- Confundir campo (propriedade das fontes) com força (depende da carga de prova).
- Dizer que capacitores em paralelo têm a mesma carga.
- Somar as cargas das duas placas e chamar o resultado de “carga armazenada”.

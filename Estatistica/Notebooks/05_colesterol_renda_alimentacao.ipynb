## Exercício 6 — Dispersão do colesterol

6 – Em um exame de colesterol, o grau médio de um grupo “A” de 150 pessoas foi de 214 mg/dl e um desvio-padrão de 22 mg/dl. Em outro grupo “B”, entretanto, o grau médio em 150 pessoas foi de 201 mg/dl e um desvio-padrão de 21 mg/dl. Em que grupo foi maior a dispersão?


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

def mostrar_totais(tabela, colunas):
    """Mostra Σ no rodapé sem acrescentar linhas ao DataFrame de cálculo."""
    rodape = {col: (tabela[col].sum() if col in colunas else '') for col in tabela.columns}
    rodape[tabela.columns[0]] = 'Σ'
    display(pd.concat([tabela, pd.DataFrame([rodape])], ignore_index=True))
```

### Comparação relativa

Como as médias são diferentes, comparamos $CV=s/\bar x$. Também apresentamos os desvios em mg/dL para distinguir a dispersão absoluta da relativa.


```python
grupos=pd.DataFrame({'Grupo':['A','B'],'n':[150,150],'Média (mg/dL)':[214,201],'Desvio (mg/dL)':[22,21]})
grupos['CV (%)']=100*grupos['Desvio (mg/dL)']/grupos['Média (mg/dL)']
display(grupos)
print('Maior desvio absoluto:',grupos.loc[grupos['Desvio (mg/dL)'].idxmax(),'Grupo'])
print('Maior dispersão relativa:',grupos.loc[grupos['CV (%)'].idxmax(),'Grupo'])
```

### Dados e tabelas individuais

As duas listas estão na ordem dos 25 pares do enunciado. Os somatórios aparecem apenas na exibição.


```python
renda=[3,5,10,10,20,20,20,30,40,50,60,70,70,80,100,100,100,120,120,140,150,180,180,200,200]
alimentacao=[1.5,2,6,7,10,12,15,8,10,20,20,25,30,25,40,35,40,30,40,40,50,40,50,60,50]
df=pd.DataFrame({'Renda':renda,'Alimentação':alimentacao})
def frequencias(v):
    t=v.value_counts().sort_index().rename_axis(v.name).reset_index(name='fi')
    t['fri']=t.fi/len(v); t['Fac']=t.fi.cumsum(); t['Fra']=t.fri.cumsum()
    return t
for col in df:
    print(col); mostrar_totais(frequencias(df[col]),['fi','fri'])
```

### Centro, dispersão, forma e quartis

O desvio usa divisor $n$ para descrever o conjunto de 25 observações. $CV=\sigma/\bar x$ permite comparar dispersões relativas. Quartis: interpolação linear do pandas.


```python
for col in df:
    v=df[col]; media=v.mean(); dp=v.std(ddof=0)
    g=((v-media)**3).mean()/dp**3
    print(f'{col}: média={media:.2f}; σ={dp:.2f}; CV={dp/media:.2%}; assimetria={g:.4f}')
    print(f'Q1={v.quantile(.25):.2f}; mediana={v.median():.2f}; Q3={v.quantile(.75):.2f}')
fig,axes=plt.subplots(1,2,figsize=(10,3))
for ax,col in zip(axes,df):
    ax.boxplot(df[col],vert=False,showmeans=True); ax.set(title=col,xlabel=col); ax.grid(True,alpha=.3)
plt.tight_layout(); plt.show()
```

### Correlação dos pares

A tabela de cálculo mantém renda e alimentação na mesma linha para cada família.


```python
contas=df.copy(); contas['X·Y']=df.Renda*df['Alimentação']; contas['X²']=df.Renda**2; contas['Y²']=df['Alimentação']**2
mostrar_totais(contas,['Renda','Alimentação','X·Y','X²','Y²'])
print(f'Pearson r = {df.Renda.corr(df["Alimentação"]):.4f}')
fig,ax=plt.subplots(figsize=(7,4)); ax.scatter(df.Renda,df['Alimentação'])
ax.set(xlabel='Renda familiar',ylabel='Gastos com alimentação',title='Renda e alimentação'); ax.grid(True,alpha=.3); plt.show()
```

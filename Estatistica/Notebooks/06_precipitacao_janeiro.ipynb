```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

def mostrar_totais(tabela, colunas):
    """Mostra Σ no rodapé sem acrescentar linhas ao DataFrame de cálculo."""
    rodape = {col: (tabela[col].sum() if col in colunas else '') for col in tabela.columns}
    rodape[tabela.columns[0]] = 'Σ'
    display(pd.concat([tabela, pd.DataFrame([rodape])], ignore_index=True))
```

### Dados

A representação individual preserva os dias sem chuva e os eventos intensos. Como há muitos valores distintos, a distribuição também pode ser resumida por classes, se necessário; os cálculos seguintes mantêm a precisão original.


```python
chuva=[0,0,0,0,4.7,0.6,17.2,1.4,11.2,1.0,3.8,0.3,0,0,0.5,7,0,0,3.3,7.6,0.9,0.3,18.2,4,4.6,22,15.6,0,3.4,0,0]
df=pd.DataFrame({'Dia':range(1,32),'Precipitação (mm)':chuva})
display(df)
```

### Distribuição por valor

O acumulado segue a ordem crescente da precipitação, e não a ordem dos dias.


```python
v=df['Precipitação (mm)']; n=len(v)
tab=v.value_counts().sort_index().rename_axis('xi (mm)').reset_index(name='fi')
tab['fri']=tab.fi/n; tab['Fac']=tab.fi.cumsum(); tab['Fra']=tab.fri.cumsum()
media=(tab['xi (mm)']*tab.fi).sum()/n
tab['xi·fi']=tab['xi (mm)']*tab.fi
tab['fi·(xi−média)²']=tab.fi*(tab['xi (mm)']-media)**2
tab['fi·(xi−média)³']=tab.fi*(tab['xi (mm)']-media)**3
mostrar_totais(tab,['fi','fri','xi·fi','fi·(xi−média)²','fi·(xi−média)³'])
```

### Medidas calculadas

Usaremos a variância **populacional** (divisor $n$), coerente com a descrição destes dados. Para dados agrupados, os pontos médios aproximam as observações:

$$\bar x=\frac{\sum f_ix_i}{n},\qquad
\sigma^2=\frac{\sum f_i(x_i-\bar x)^2}{n},\qquad
\sigma=\sqrt{\sigma^2},\qquad
g_1=\frac{\sum f_i(x_i-\bar x)^3}{n\sigma^3}.$$

O coeficiente de variação é $CV=\sigma/\bar x$ (em porcentagem, multiplicar por 100).


```python
var=tab['fi·(xi−média)²'].sum()/n; dp=np.sqrt(var)
g=tab['fi·(xi−média)³'].sum()/(n*dp**3)
print(f'Média={media:.3f} mm; mediana={v.median():.3f} mm; moda(s)={v.mode().tolist()}')
print(f'Variância={var:.3f} mm²; desvio={dp:.3f} mm; CV={dp/media:.2%}; assimetria={g:.3f}')
print(f'Q1={v.quantile(.25):.3f}; Q2={v.quantile(.5):.3f}; Q3={v.quantile(.75):.3f} mm')
print(f'Dias sem chuva={(v==0).sum()} de {n}')
```

### Gráficos

A série por dia mostra quando choveu; o acumulado responde à distribuição das intensidades.


```python
fig,axes=plt.subplots(1,2,figsize=(11,4))
axes[0].bar(df.Dia,v); axes[0].set(xlabel='Dia',ylabel='Precipitação (mm)',title='Janeiro de 2014')
axes[1].step(tab['xi (mm)'],tab.Fra,where='post'); axes[1].set(xlabel='Precipitação (mm)',ylabel='Frequência relativa acumulada',title='Distribuição acumulada',ylim=(0,1.05))
for ax in axes: ax.grid(True,alpha=.3)
plt.tight_layout(); plt.show()
fig,ax=plt.subplots(figsize=(7,2.5)); ax.boxplot(v,vert=False,showmeans=True); ax.set(xlabel='Precipitação (mm)',title='Boxplot'); ax.grid(True,alpha=.3); plt.show()
```

**Interpretação:** muitos zeros tornam a distribuição assimétrica à direita; a média é puxada pelos dias com chuva intensa. Os quartis seguem a convenção de interpolação linear do pandas.

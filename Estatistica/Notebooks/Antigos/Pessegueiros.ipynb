# 1. 40 Pessegueiros produzem as seguintes quantidades por árvore em caixas.

2. Os dados foram colocadas em um array de uma dimensão

```python
import numpy as np
import matplotlib.pyplot as plt
arvores = np.array([11.1, 21., 4.4, 18.5, 10.7, 12.2, 23.5, 19.1, 26.2, 3.2, 12.5, 16.4, 6.1, 16.4, 15.8, 12.6, 14.8, 7.4,  3.5, 8.1,
32.4, 11.2, 27.5, 15.1, 25., 4.7, 22.6, 9.2, 16.2, 2.9, 7.8, 22.3, 32.8, 6., 18.2, 13.7, 16., 10., 14.5, 19.1])
print(f"\nO array árvores é \n{arvores}")

print(f"\nA média de caixas por arvore é: {round(arvores.mean(),2)}")
print(f"\n A árvore que mais produziu, deu: {(arvores.max())} caixas")
print(f"\n A árvore que menos produziu, deu: {(arvores.min())} caixas")
print(f"\n A mediana de caixas produzidas é: {np.median(arvores)}")
print(f"\nO primeiro quartil é: {round(np.quantile(arvores, 0.25), 2)}")
print(f"\n O terceiro Quartil é: {np.quantile(arvores, 0.75)}")
plt.boxplot(arvores, orientation='horizontal');
plt.show()
```

## Pela regra de Sturges:

Pela regra de Sturges, o número de classes é dado por:
$$K=1+3.3\cdot\log_2(n)$$

```python
n= len(arvores)
k = int(np.ceil(1 + np.log2(n)))
print(k)
```

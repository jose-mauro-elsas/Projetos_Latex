## Desenvolvimento da fórmula da variância

Para dados agrupados em classes, a variância pode ser calculada por:

$$
s^2 =
\frac{\displaystyle\sum_{i=1}^{k}
f_i\left(x_i-\bar{x}\right)^2}{n}
$$

em que:

- $x_i$ é o valor representativo da classe, normalmente o centro da classe;
- $f_i$ é a frequência absoluta da classe;
- $k$ é o número de classes;
- $n$ é o número total de observações;
- $\bar{x}$ é a média dos dados agrupados.

O número total de observações corresponde à soma das frequências:

$$
n = \sum_{i=1}^{k} f_i
$$

A média dos dados agrupados é dada por:

$$
\bar{x}
=
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i}{n}
$$

Desenvolvimento do quadrado da diferença

Utilizando a identidade algébrica:

$$
(a-b)^2 = a^2-2ab+b^2
$$

temos:

$$
\left(x_i-\bar{x}\right)^2
=
x_i^2-2x_i\bar{x}+\bar{x}^2
$$

Substituindo essa expressão na fórmula da variância:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}
f_i\left(
x_i^2-2x_i\bar{x}+\bar{x}^2
\right)
}{n}
$$

Distribuindo a frequência $f_i$:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}
\left(
f_i x_i^2
-
2f_i x_i\bar{x}
+
f_i\bar{x}^2
\right)
}{n}
$$

Separando os somatórios:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}f_i x_i^2
-
2\bar{x}\displaystyle\sum_{i=1}^{k}f_i x_i
+
\bar{x}^2\displaystyle\sum_{i=1}^{k}f_i
}{n}
$$

Como:

$$
\bar{x}
=
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i}{n}
$$

podemos escrever:

$$
\sum_{i=1}^{k}f_i x_i
=
n\bar{x}
$$

Além disso:

$$
\sum_{i=1}^{k}f_i=n
$$

Substituindo essas duas relações na expressão da variância:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}f_i x_i^2
-
2\bar{x}\left(n\bar{x}\right)
+
\bar{x}^2n
}{n}
$$

Simplificando:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}f_i x_i^2
-
2n\bar{x}^2
+
n\bar{x}^2
}{n}
$$

Como:

$$
-2n\bar{x}^2+n\bar{x}^2
=
-n\bar{x}^2
$$

obtemos:

$$
s^2 =
\frac{
\displaystyle\sum_{i=1}^{k}f_i x_i^2
-
n\bar{x}^2
}{n}
$$

Separando a divisão por $n$:

$$
\boxed{
s^2 =
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i^2}{n}
-
\bar{x}^2
}
$$

Como a média agrupada é:

$$
\bar{x}
=
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i}{n}
$$

a fórmula também pode ser escrita como:

$$
\boxed{
s^2 =
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i^2}{n}
-
\left(
\frac{\displaystyle\sum_{i=1}^{k}f_i x_i}{n}
\right)^2
}
$$

Essa expressão é conhecida como **fórmula desenvolvida** ou **fórmula computacional da variância**.


```python
df["fi_xi"] = df["fi"] * df["xi"]

df["fi_xi2"] = (
    df["fi"] * df["xi"] ** 2
)

n = df["fi"].sum()

media_agrupada = df["fi_xi"].sum() / n

variancia_agrupada = (
    df["fi_xi2"].sum() / n
    - media_agrupada ** 2
)

print(f"Média agrupada: {media_agrupada:.2f}")
print(f"Variância agrupada: {variancia_agrupada:.2f}")
```

O desvio padrão aproximado é dado por:

$$s = \sqrt{s^2}$$

```python
# Desvio padrão aproximado
desvio_padrao_agrupado = np.sqrt(variancia_agrupada)

print(f"O Desvio Padrão aproximado: {desvio_padrao_agrupado.round(2)} caixas")
```

## 5. Medidas de tendência central.
### 5.1 Moda.

A moda é uma medida de posição e de tendência central que descreve o valor mais frequente ou classe mais frequente de um conjunto de observações, ou seja, o valor de maior ocorrência dentre os valores observados. Podemos usar a frequência relativa, uma vez que as duas são proporcionais o valor da moda será o mesmo.

Quando os dados estão agrupados em classes de mesma amplitude, a moda será o valor dominante da classe que apresenta a maior frequência, a qual é denominada classe modal. Como já vimos, a amplitude de classe é a diferença entre os limites superior e inferior de uma determinada classe. Assim, quando as amplitudes são todas iguais, a moda estará contida na classe de maior frequência. A seguir, veremos os principais métodos empregados no cálculo da moda de distribuições agrupadas por intervalos de classe: moda bruta, moda de Pearson, moda de Czuber e moda de King.

### 5.1 Média

 A média aritmética simples está muito presente em nosso cotidiano, seja no consumo médio de combustível, na temperatura média ou na renda per capita. Essa medida é definida como o **QUOCIENTE** entre a **SOMA DE TODOS OS ELEMENTOS** e o **NÚMERO DELES**. A propriedade principal da média é preservar a soma dos elementos de um conjunto de dados.

Podemos adotar o seguinte raciocínio para encontrarmos a fórmula da média aritmética. Dada uma lista de n números, $(x_1, x_2, \dots, x_n)$ a soma de seus termos é igual a:

$$
    \overbrace{x_1+x_2+x_3+\dots+x_n}^{\text{n fatores}}
$$
A média aritmética dessa lista é um número , tal que, se todos os elementos forem substituídos por \(\bar{x}\), a soma da lista permanecerá preservada. Assim, substituindo todos os elementos por $\bar{x}, \bar{x}, \dots, \bar{x}$, teremos uma nova lista, cuja soma é:

$$
    \overbrace{\bar{x}+\bar{x}+\dots+\bar{x}}^{\text{n fatores}}=n\times\bar{x}
$$
Como as somas das duas listas são iguais, temos:
$$
    n\times\bar{x}=\bar{x_1}+\bar{x_2}+\dots+\bar{x_n}
$$

Portanto, a média aritmética é:

$$
    \bar{x}=\frac{{x_1}+{x_2}+\dots+x_n}{n}
$$

Como já foi visto anteriormente, a média aproximada dos dados agrupados em classes (utilizando o Ponto médio da classe)é dado por:

$$\bar x = \frac{\sum_{i=1}^n (x_i\cdot f_i)}{n}$$

**Aqui a frequência é a frequência simples**

### 5.2 Mediana

A mediana é, simultaneamente, uma MEDIDA DE POSIÇÃO, de TENDÊNCIA CENTRAL e SEPARATRIZ. Ela caracteriza a posição central de uma série de valores. Além disso, também separa uma série de valores em duas partes de tamanhos iguais, cada uma contendo o mesmo número de elementos. Muitas vezes, a mediana é designada como valor mediano, sendo representada pelos símbolos $M_d$ ou, em menor frequência, $\tilde{x}$

Mediana para Dados não Agrupados

O método para determinação da mediana envolve a realização de uma etapa anterior, que consiste na ordenação do conjunto de dados. Feito isso, a mediana é o elemento que ocupa a **POSIÇÃO CENTRAL** de uma série de observações **ORDENADA** segundo suas grandezas (isto é, dados brutos organizados em rol crescente ou decrescente).

Por exemplo, vamos determinar a mediana da seguinte série de valores:

$$
    (3, 17, 13, 19, 2, 5, 7, 1, 8, 21, 9)
$$

```python
lista = [3, 17, 13, 19, 2, 5, 7, 1, 8, 21, 9]
listarr = np.array(lista)
listarr.sort()
print(f"O conjunto ordenado é: {listarr}")
```

Quando o número de elementos do conjunto é ímpar:

O conjunto de dados ordenado tem 11 elementos, portanto o elemento que têm 50% do elementos antes e depois dele simultaneamente é aquele que ocupa a sexta posição. O elemento da sexta posição é o 8.

Quando o número de elementos do conjunto é par, teremos um par de elementos centrais e a mediana seria a média aritmética desses dois valores. Se a lista original fosse:

```python
lista_par = [3, 17, 13, 19, 2, 5, 7, 1, 8, 9]
listarp = np.array(lista_par)
listarp.sort()
print(f"O conjunto ordenado é: {listarp}")
```

Assim a mediana no conjunto de 10 valores seria a média dos valores centrais

```python
mediana = (listarp[4] + listarp[5])/2
print(f"O valor da Mediana do conjunto é: {mediana}")
```

### Coeficiente de Variação (CV)

O coeficiente de variação mede a dispersão dos dados em relação à média. Quanto mais próximo de zero, menor é a variabilidade relativa e mais concentrados estão os valores em torno da média. Quanto maior o coeficiente, maior é a dispersão relativa dos dados.

Também é importante que a média seja diferente de zero e, em geral, positiva. Quando a média está muito próxima de zero, o coeficiente de variação pode ficar muito grande e perder utilidade interpretativa.

Um coeficiente de variação de 0,48 significa que o desvio padrão corresponde a 48% da média.

```python
s = desvio_padrao_agrupado
med = media_agrupada
cv = s/med
print(f"O coeficiente de variação da Produção é: {cv.round(2)}")
```
